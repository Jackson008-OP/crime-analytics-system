```python
# Load CSV file
df = pd.read_csv("C:/Users/ASUS/Desktop/Crime Analytics/Dataset/crime_data.csv")

# Show first 5 rows
df.head()
```

```python
# Remove leading/trailing spaces
df.columns = df.columns.str.strip()

# Replace spaces with underscores
df.columns = df.columns.str.replace(" ", "_")

# Convert to uppercase for consistency
df.columns = df.columns.str.upper()

df.columns
```

```python
df.isnull().sum()
```

```python
df = df.fillna(0)
```

```python
df["YEAR"] = df["YEAR"].astype(int)
```

```python
df = df[df["STATE/UT"] != "India"]
```

```python
df.to_csv("../Dataset/cleaned_crime_data.csv", index=False)
```
